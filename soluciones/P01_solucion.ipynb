## 1. Inspección inicial · 10 minutos

Presentar la cantidad de filas y columnas, los nombres de las columnas, sus tipos y cinco registros. Explicar qué representa una fila y por qué la lectura inicial conserva los campos como texto.

📘 **Apoyo:** `shape`, `columns`, `dtypes`, `head` e `info`.

✅ **Comprobación:** El corte del manifiesto contiene 92.738 filas y 32 columnas. La tabla original debe conservar esa cantidad de filas durante el ejercicio.

````python
# Desarrollo de la actividad 1
# Mostrar dimensiones, tipos y primeros registros de suelo.
# Guardar el total de filas original para los controles posteriores.
total_filas_orig = len(suelo)
print(f"Total de filas originales: {total_filas_orig}")
print(f"Dimensiones (filas, columnas): {suelo.shape}")
print("\nNombres de las columnas:")
print(suelo.columns.tolist())
print("\nTipos de datos iniciales:")
print(suelo.dtypes)
print("\nPrimeros 5 registros:")
display(suelo.head())
suelo.info(memory_usage="deep")
assert total_filas_orig == 92738, f"Se esperaban 92,738 filas pero se obtuvieron {total_filas_orig}"
````

## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.

campos_interes = ["Secuencial", "Fecha de Análisis", "Departamento", "Municipio", "pH agua:suelo"]

registros_perfil = []
for col in campos_interes:
    s = suelo[col]
    nulos_pandas = s.isna().sum()
    textos_vacios = (s.str.strip() == "").sum()
    marcadores_nd = (s.str.strip().str.upper() == "ND").sum()
    
    registros_perfil.append({
        "campo": col,
        "nulos_pandas": nulos_pandas,
        "textos_vacios": textos_vacios,
        "marcadores_ND": marcadores_nd,
      
    })

perfil = pd.DataFrame(registros_perfil)
display(perfil)
````

**Interpretación del resultado:**

Se evidencioa que la tabla perfil refleja que, al haber configurado keep_default_na=False en la lectura, la columna nulos_pandas es cero para los cinco campos. En su lugar, los valores faltantes o no registrados se encuentran adecuadamente discriminados entre textos_vacios y marcadores explícitos "ND".

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
# Mostrar los fallos de conversión y sus valores originales.

suelo["ph"] = pd.to_numeric(suelo["pH agua:suelo"], errors="coerce")
suelo["fecha_dt"] = pd.to_datetime(suelo["Fecha de Análisis"], format="%d/%m/%Y", errors="coerce")

ph_orig = suelo["pH agua:suelo"].str.strip()
ph_vacio_nd = (ph_orig == "") | (ph_orig.str.upper() == "ND")
ph_fallo_formato = suelo["ph"].isna() & (~ph_vacio_nd)

print(f"pH - Sin formato numérico válido: {suelo['ph'].isna().sum()}")
print(f"  └─ De los cuales son vacíos o ND: {ph_vacio_nd.sum()}")
print(f"  └─ De los cuales son fallos de formato: {ph_fallo_formato.sum()}")

if ph_fallo_formato.sum() > 0:
    print("\nEjemplos de texto con fallo de formato en pH:")
    display(suelo.loc[ph_fallo_formato, ["pH agua:suelo"]].head(5))

fecha_orig = suelo["Fecha de Análisis"].str.strip()
fecha_vacia_nd = (fecha_orig == "") | (fecha_orig.str.upper() == "ND")
fecha_fallo_formato = suelo["fecha_dt"].isna() & (~fecha_vacia_nd)

print(f"\nFecha - Sin fecha válida: {suelo['fecha_dt'].isna().sum()}")
print(f"  └─ De los cuales son vacíos o ND: {fecha_vacia_nd.sum()}")
print(f"  └─ De los cuales son fallos de formato: {fecha_fallo_formato.sum()}")

if fecha_fallo_formato.sum() > 0:
    print("\nEjemplos de texto con fallo de formato en Fecha:")
    display(suelo.loc[fecha_fallo_formato, ["Fecha de Análisis"]].head(5))
````

**Interpretación del resultado:**
La conversión mediante errors="coerce" permitió transformar los registros válidos a números reales (ph) y objetos de fecha (fecha_dt). Los datos no convertidos corresponden principalmente a ausencias de origen, campos vacíos o "ND", evitando imputar erróneamente un valor cero a registros que realmente carecen de medición.

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
# Comprobar la conservación del total original con assert.
motivo_ph_ausente = suelo["ph"].isna()
motivo_ph_fuera_rango = ~suelo["ph"].between(0, 14) & (~motivo_ph_ausente)
motivo_fecha_no_interpretable = suelo["fecha_dt"].isna()

suelo["en_revision"] = motivo_ph_ausente | motivo_ph_fuera_rango | motivo_fecha_no_interpretable

print("Conteos por motivo individual:")
print(f" - pH ausente: {motivo_ph_ausente.sum()}")
print(f" - pH fuera del rango [0, 14]: {motivo_ph_fuera_rango.sum()}")
print(f" - Fecha no interpretable: {motivo_fecha_no_interpretable.sum()}")

print("\nFrecuencia total en revisión:")
print(suelo["en_revision"].value_counts())

total_rev = suelo["en_revision"].sum()
total_ok = (~suelo["en_revision"]).sum()
assert (total_rev + total_ok) == total_filas_orig, "Error: El número total de filas cambió."
print(f"\n✅ Verificación exitosa: Filas en revisión ({total_rev}) + Sin revisión ({total_ok}) = {total_filas_orig} filas originales.")
````

**Interpretación del resultado:**

Se aprecia que la columna booleana en_revision permite identificar y auditar las filas que contienen inconsistencias sin necesidad de eliminarlas del dataset principal. La suma del total de filas revisadas y no revisadas coincide exactamente con el conteo original de 92.738 registros.

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
# Redactar los dos hallazgos en la celda Markdown siguiente.
datos_controles = [
    {"control": "total_entrada", "cantidad": total_filas_orig},
    {"control": "ph_sin_numero", "cantidad": motivo_ph_ausente.sum()},
    {"control": "ph_fuera_intervalo", "cantidad": motivo_ph_fuera_rango.sum()},
    {"control": "fecha_no_interpretable", "cantidad": motivo_fecha_no_interpretable.sum()},
    {"control": "total_filas_en_revision", "cantidad": suelo["en_revision"].sum()}
]

controles = pd.DataFrame(datos_controles)
display(controles)

ruta_salida = OUT / "controles_agrosavia.csv"
controles.to_csv(ruta_salida, index=False, encoding="utf-8-sig")
assert ruta_salida.is_file()
print("Archivo generado:", ruta_salida)

print(f"✅ Archivo exportado exitosamente en: {ruta_salida}")
````

**Interpretación del resultado:**

Hallazgo 1: La mayoría de los registros marcados para revisión se deben a la ausencia del dato de pH (ph_sin_numero), lo que es un porcentaje muy bajo de error, lo que permite que al excluir esas filas practicamente no cambia el tamaño de la muestra.

Hallazgo 2: La cantidad de datos con ph_fuera_intervalo es marginal, lo que muestra que cuando el parámetro numérico de pH existe, generalmente respeta la escala física de 0 a 14.
